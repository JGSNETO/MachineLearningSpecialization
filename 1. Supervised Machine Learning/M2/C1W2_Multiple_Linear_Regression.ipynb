# Multiple Linear Regression and Vectorization

## Overview

Multiple linear regression extends linear regression by using **multiple input features** to predict a target variable.

Instead of predicting an output from a single feature, the model combines several features, each with its own parameter. This introduces vector notation and makes **vectorization** important for efficient implementation.

The main topics covered are:

- Multiple linear regression
- Model representation with multiple features
- Vector and dot-product notation
- Vectorized implementation with NumPy
- Vectorized gradient descent
- The normal equation as an alternative optimization method

---

## 1. Multiple Linear Regression

In single-feature linear regression, the prediction depends on one input feature.

Multiple linear regression extends this concept by using several features:

- `X₁` — first feature
- `X₂` — second feature
- ...
- `Xₙ` — nth feature

For example, house price prediction could use:

- House size
- Number of bedrooms
- Number of floors
- House age

Each training example therefore contains a **vector of features** rather than a single value.

---

## 2. Model Representation

The prediction is calculated as a weighted combination of all features plus a bias:

\[
f_{w,b}(X) = w_1X_1 + w_2X_2 + \dots + w_nX_n + b
\]

Each parameter `w_j` corresponds to one feature.

The weight determines how strongly that feature contributes to the prediction, while `b` provides the bias term.

### Vector notation

The individual weights can be grouped into a vector:

\[
W =
\begin{bmatrix}
w_1 \\
w_2 \\
\vdots \\
w_n
\end{bmatrix}
\]

The features can similarly be represented as:

\[
X =
\begin{bmatrix}
X_1 \\
X_2 \\
\vdots \\
X_n
\end{bmatrix}
\]

The model can then be expressed using the dot product:

\[
f_{W,b}(X) = W \cdot X + b
\]

The dot product performs the sum of the corresponding weight-feature products.

This representation becomes especially useful when the number of features increases.

---

## 3. Why Vectorization?

A straightforward implementation could calculate every feature contribution individually:

```python
prediction = w1 * x1 + w2 * x2 + w3 * x3
```

For many features, this approach becomes impractical.

A loop can make the implementation more general:

```python
for j in range(n):
    prediction += w[j] * x[j]
```

However, the calculations are still performed sequentially.

**Vectorization** replaces these explicit loops with operations on entire arrays or vectors.

For example, NumPy can calculate the dot product directly:

```python
prediction = np.dot(w, x) + b
```

The mathematical operation and the implementation now closely match:

\[
W \cdot X + b
\]

---

## 4. Why Vectorization Is Faster

Vectorized operations can take advantage of optimized numerical libraries and specialized hardware.

Instead of explicitly processing one element at a time:

```text
x₁ → calculation
x₂ → calculation
x₃ → calculation
...
xₙ → calculation
```

vectorized operations allow the underlying implementation to process many calculations efficiently.

This can take advantage of:

- CPU parallelism
- GPU acceleration
- Optimized numerical libraries
- Efficient array operations

The performance advantage becomes increasingly important as the number of features and training examples grows.

---

## 5. Vectorization in Multiple Linear Regression

Vectorization is particularly useful for multiple linear regression because the model contains many parameters.

Instead of individually calculating:

```text
w₁ × x₁
w₂ × x₂
w₃ × x₃
...
wₙ × xₙ
```

the complete calculation can be expressed as:

```python
np.dot(w, x)
```

The resulting prediction is:

```python
prediction = np.dot(w, x) + b
```

This provides a concise and efficient implementation of the multiple-feature model.

---

## 6. Vectorized Gradient Descent

Gradient descent can also be extended from one feature to multiple features.

The parameters are represented collectively as the vector `w`:

```text
w = [w₁, w₂, ..., wₙ]
```

Each parameter `w_j` receives an update based on the corresponding feature and the prediction error.

Conceptually:

\[
w_j \leftarrow w_j - \alpha
\frac{\partial J}{\partial w_j}
\]

while the bias is updated separately:

\[
b \leftarrow b - \alpha
\frac{\partial J}{\partial b}
\]

The important extension is that there are now **multiple weight parameters**, one for each feature.

Vectorization allows these operations to be implemented using array operations rather than manually updating each parameter through separate code.

This makes the implementation much more scalable for models containing many features.

---

## 7. NumPy and Machine Learning

NumPy provides the array-based operations required for efficient vectorized implementations.

Important operations include:

```python
import numpy as np

prediction = np.dot(w, x) + b
```

Instead of explicitly implementing loops, the numerical work is delegated to optimized array operations.

This programming approach is widely useful beyond linear regression and forms an important foundation for implementing machine learning algorithms efficiently.

---

## 8. Normal Equation

The **normal equation** provides an alternative way to find the parameters for linear regression.

Instead of iteratively updating the parameters with gradient descent, the normal equation uses linear algebra to solve for the optimal parameters directly.

### Gradient descent

```text
Initialize parameters
        ↓
Calculate gradients
        ↓
Update parameters
        ↓
Repeat
        ↓
Converged solution
```

### Normal equation

```text
Training data
      ↓
Linear algebra calculation
      ↓
Direct solution for parameters
```

The normal equation therefore does not require iterative gradient-descent updates.

---

## 9. Gradient Descent vs. Normal Equation

| Aspect | Gradient Descent | Normal Equation |
|---|---|---|
| Optimization | Iterative | Direct |
| Main approach | Repeated parameter updates | Linear algebra |
| Requires learning rate | Yes | No |
| Applicable beyond linear regression | Yes | No |
| Large feature sets | Generally preferred | Can become slow |
| Implementation | General optimization approach | Specific to linear regression |

The normal equation can be useful for linear regression, but it does not generalize to algorithms such as logistic regression and neural networks.

For practical machine-learning implementations, gradient descent provides a more general and scalable approach.

---

## 10. From Mathematical Model to Efficient Implementation

The progression introduced in this topic can be viewed as:

```text
Multiple Features
       ↓
Multiple Parameters
       ↓
Vector Representation
       ↓
Dot Product
       ↓
Vectorized Implementation
       ↓
Vectorized Gradient Descent
       ↓
Efficient Machine Learning Computation
```

The key idea is that mathematical vector notation is not only useful for expressing the model more compactly—it also leads directly to more efficient implementations.

## 11. Practical Takeaways

- The weights and features can be represented as vectors.

- **Vectorization** replaces explicit loops with efficient array operations.
- NumPy provides the core operations needed for vectorized implementations.

## 12. Key Concept

The central connection in this topic is:

> **Vector notation makes multiple linear regression mathematically compact, while vectorization makes its implementation computationally efficient.**

The model moves from handling individual feature-weight calculations to treating the parameters and features as vectors, allowing optimized numerical operations to perform the underlying computations efficiently.

````python
'''
1. Vectorized Multiple Linear Regression Prediction
'''
import numpy as np

# Features:
# [house_size, bedrooms, floors, age]
x = np.array([2100, 4, 2, 10])

# Model parameters
w = np.array([150, 20000, 10000, -1000])
b = 50000

# Vectorized prediction
prediction = np.dot(w, x) + b

print(f"Predicted price: ${prediction:,.2f}")
````

````python
'''
2. Vectorized Gradient Descent for Multiple Linear Regression
This example shows how the gradients for all weights can be calculated simultaneously without looping through the features.
'''
import numpy as np


def compute_gradient(X, y, w, b):
    """
    Compute gradients for multiple linear regression.

    X: training examples, shape (m, n)
    y: target values, shape (m,)
    w: parameter vector, shape (n,)
    b: bias
    """

    m = X.shape[0]

    # Vectorized predictions for all training examples
    predictions = X @ w + b

    # Prediction errors
    errors = predictions - y

    # Vectorized gradients
    dj_dw = (X.T @ errors) / m
    dj_db = np.sum(errors) / m

    return dj_dw, dj_db


# Training data
X = np.array([
    [2100, 4, 2, 10],
    [1600, 3, 1, 15],
    [2400, 4, 2, 5],
    [1400, 2, 1, 20]
])

y = np.array([
    400000,
    300000,
    500000,
    250000
])

# Initialize parameters
w = np.zeros(X.shape[1])
b = 0

learning_rate = 0.001
iterations = 1000

# Gradient descent
for _ in range(iterations):

    dj_dw, dj_db = compute_gradient(X, y, w, b)

    # Update all weights simultaneously
    w -= learning_rate * dj_dw

    # Update bias
    b -= learning_rate * dj_db

print("Learned weights:", w)
print("Learned bias:", b)
````
