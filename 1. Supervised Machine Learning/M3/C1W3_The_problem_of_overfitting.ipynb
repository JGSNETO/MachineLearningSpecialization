# Overfitting, Underfitting & Regularization

Machine learning models need to learn the underlying patterns in data without becoming either too simple or unnecessarily complex. **Underfitting** and **overfitting** are two common problems that affect a model's ability to generalize to new data.

## 1. Underfitting vs. Overfitting

### Underfitting

**Underfitting** occurs when a model is too simple to capture the underlying pattern in the training data.

- Performs poorly even on the training set.
- Fails to capture important relationships between features and the target.
- Associated with **high bias**.
- Usually caused by an overly simple model or insufficiently informative features.

For example, a simple linear model may underfit housing-price data when the actual relationship between the features and price is nonlinear.

### Just-Right Fit

A well-designed model captures the important patterns without fitting random fluctuations in the training data.

- Performs well on the training data.
- Generalizes well to unseen data.
- Balances model complexity and predictive performance.

For example, a quadratic model may capture a nonlinear relationship better than a simple linear model while still generalizing effectively.

### Overfitting

**Overfitting** occurs when a model fits the training data too closely, including noise and random fluctuations.

- Can achieve extremely good or even perfect training performance.
- Performs poorly on new, unseen data.
- Associated with **high variance**.
- Usually caused by an excessively complex model relative to the amount or quality of training data.

A high-order polynomial can illustrate this problem by following almost every training point while behaving erratically outside the training examples.

### Bias and Variance

| Problem | Model Complexity | Training Performance | Generalization | Main Characteristic |
|---|---|---|---|---|
| Underfitting | Too simple | Poor | Poor | High bias |
| Good fit | Appropriate | Good | Good | Balanced |
| Overfitting | Too complex | Very good | Poor | High variance |

The overall goal is to find a model that **generalizes well to new data**, rather than simply minimizing training error.

---

## 2. Examples in Regression and Classification

### Linear Regression

Model complexity can be increased by adding polynomial features:

- **Linear model:** May be too simple → underfitting.
- **Quadratic model:** Can capture additional nonlinear structure → potentially a good fit.
- **High-order polynomial:** Can follow training data extremely closely → overfitting.

### Logistic Regression

The same concept applies to classification:

- A **linear decision boundary** may be too simple for complex classification patterns.
- Adding **quadratic features** can provide a more suitable decision boundary.
- Very high-order polynomial features can produce an overly complex boundary that perfectly separates training examples but generalizes poorly.

Therefore, increasing model complexity does not automatically improve performance. The model must have enough complexity to capture meaningful patterns without learning noise.

---

## 3. How to Address Overfitting

Several approaches can help reduce overfitting.

### Collect More Training Data

More representative training data can help the model identify the underlying pattern instead of fitting individual examples or noise.

This can improve generalization and reduce the tendency to fit overly complex patterns.

### Feature Selection

Another approach is to reduce the number of features used by the model.

**Feature selection** involves identifying and retaining the most useful features while removing less relevant ones.

Advantages:

- Reduces model complexity.
- Can reduce overfitting.
- May improve computational efficiency.

However, removing features has a trade-off: a discarded feature may contain useful information. Feature selection should therefore be performed carefully.

### Regularization

Regularization provides another way to control model complexity without necessarily removing features.

Instead of eliminating parameters, regularization **penalizes large parameter values**, encouraging the model to use smaller weights.

This allows the model to retain all features while reducing the influence of individual features.

---

## 4. Regularization

Regularization modifies the model's **cost function** by adding a penalty for large parameter values.

For a model with parameters `w₁, w₂, ..., wₙ`, the regularization term is based on the squared parameter values:

```text
Regularization penalty ∝ λ(w₁² + w₂² + ... + wₙ²)
```

The resulting cost function therefore combines two objectives:

1. Fit the training data well.
2. Keep the model parameters small.

Conceptually:

```text
Regularized Cost
      =
Training Error
      +
Complexity Penalty
```

The complexity penalty discourages the model from assigning excessively large weights to features.

---

## 5. The Regularization Parameter λ

The parameter **λ (lambda)** controls how strongly regularization is applied.

### λ = 0

No regularization is applied.

```text
λ = 0 → No penalty
```

The model is free to choose large parameter values, which can increase the risk of overfitting.

### Very Large λ

A very large regularization parameter heavily penalizes large weights.

```text
Large λ → Very small parameters
```

If the penalty becomes too strong, the model can become excessively simple and **underfit** the data.

### Appropriate λ

The objective is to choose a value that provides a useful balance:

```text
Too little regularization → Overfitting
        ↓
   Appropriate λ
        ↓
Good generalization
        ↓
Too much regularization → Underfitting
```

Therefore, λ controls the trade-off between fitting the training data and keeping the model simple.

---

## 6. Regularization vs. Feature Selection

These approaches control model complexity in different ways.

| Approach | What happens to features? |
|---|---|
| Feature selection | Some features are removed |
| Regularization | Features are retained, but their parameter values are reduced |

Regularization is useful when many features may contain valuable information but allowing unrestricted parameter values could lead to overfitting.

It is also widely applicable to different machine learning algorithms, including **neural networks**.

---

## 7. Bias Term

Regularization is typically applied to the model's feature parameters but **not to the bias/intercept term**.

The purpose is to control the influence of the input features without unnecessarily restricting the model's baseline prediction.

In practice, the regularization penalty is therefore applied to the weights while excluding the bias parameter.

---

## 8. Why Regularization Works

A model with very large parameter values can produce highly complex behavior that closely follows the training data.

Regularization discourages this behavior by making large parameters costly.

```text
Large parameters
       ↓
Higher regularization penalty
       ↓
Optimization favors smaller parameters
       ↓
Simpler model
       ↓
Better generalization
```

Regularization therefore provides a systematic way to control model complexity without manually deciding which individual features to remove.

---

## Key Takeaways

- **Underfitting** occurs when a model is too simple and has **high bias**.
- **Overfitting** occurs when a model is too complex and has **high variance**.
- The goal is to find a model that **generalizes well** to unseen data.
- More training data can help reduce overfitting.
- **Feature selection** reduces model complexity by removing features.
- **Regularization** reduces model complexity by penalizing large parameter values.
- Regularization modifies the **cost function** by adding a penalty proportional to the squared weights.
- **λ (lambda)** controls the strength of regularization.
- `λ = 0` means no regularization.
- Excessively large λ can cause underfitting.
- Regularization usually applies to the **weights, not the bias term**.
- Unlike feature selection, regularization allows the model to retain all features while controlling their influence.
- The fundamental objective is to balance model complexity and training performance to achieve **good generalization**.

# Regularization in Linear and Logistic Regression

Regularization is a technique used in supervised machine learning to reduce overfitting by penalizing large model parameters. It modifies the cost function and gradient descent updates to encourage simpler models that generalize better to unseen data.

## 1. Regularization in Linear Regression

In linear regression, the model predicts a continuous value using a weighted combination of input features:

\[
f_{\mathbf{w},b}(\mathbf{x})=\mathbf{w}\cdot\mathbf{x}+b
\]

When a model has many features or limited training data, it may learn excessively large weights and fit noise in the training set. Regularization helps control this behavior.

### Regularized Cost Function

The standard squared-error cost function is extended with a penalty proportional to the sum of squared weights:

\[
J(\mathbf{w},b)=
\frac{1}{2m}\sum_{i=1}^{m}
\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})-y^{(i)}\right)^2
+\frac{\lambda}{2m}\sum_{j=1}^{n}w_j^2
\]

Where:

- \(m\): number of training examples.
- \(n\): number of input features.
- \(w_j\): weight associated with feature \(j\).
- \(b\): bias or intercept.
- \(\lambda\): regularization parameter.

The cost function combines two objectives:

1. Minimize prediction errors on the training data.
2. Penalize large weights to control model complexity.

The regularization term is scaled by \(1/(2m)\), making the parameter \(\lambda\) easier to tune relative to the training-error term.

### Gradient Descent Updates

Regularization introduces an additional term into the gradient descent update for each weight.

For each feature weight \(w_j\):

\[
w_j := w_j-\alpha\left[
\frac{1}{m}\sum_{i=1}^{m}
\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})-y^{(i)}\right)x_j^{(i)}
+\frac{\lambda}{m}w_j
\right]
\]

The bias update remains unchanged:

\[
b := b-\alpha\frac{1}{m}
\sum_{i=1}^{m}
\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})-y^{(i)}\right)
\]

Where \(\alpha\) is the learning rate.

### Intuition Behind Weight Shrinkage

The weight update can be rewritten as:

\[
w_j := w_j\left(1-\frac{\alpha\lambda}{m}\right)
-\alpha\frac{1}{m}\sum_{i=1}^{m}
\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})-y^{(i)}\right)x_j^{(i)}
\]

The factor \(1-\frac{\alpha\lambda}{m}\) gradually shrinks the existing weight, while the usual gradient term adjusts it to improve predictions.

This shrinkage discourages excessively large weights and helps prevent overfitting. The bias is not regularized.

## 2. Regularization in Logistic Regression

Logistic regression predicts probabilities for classification tasks. Its model uses the sigmoid function:

\[
f_{\mathbf{w},b}(\mathbf{x})=
\frac{1}{1+e^{-(\mathbf{w}\cdot\mathbf{x}+b)}}
\]

Overfitting can occur when logistic regression uses many features or high-order polynomial features. The resulting decision boundary may become excessively complex and fit the training examples too closely.

Regularization addresses this problem by penalizing large weights, helping produce simpler decision boundaries that generalize better.

### Regularized Cost Function

The standard logistic regression cost function uses the binary cross-entropy loss. Regularization adds a squared-weight penalty:

\[
\begin{aligned}
J(\mathbf{w},b)={}&-\frac{1}{m}\sum_{i=1}^{m}
\left[
y^{(i)}\log f_{\mathbf{w},b}(\mathbf{x}^{(i)})\right.\\
&\left.+(1-y^{(i)})\log
\left(1-f_{\mathbf{w},b}(\mathbf{x}^{(i)})\right)
\right]\\
&+\frac{\lambda}{2m}\sum_{j=1}^{n}w_j^2
\end{aligned}
\]

The first term measures classification prediction error, while the second discourages large weights.

### Gradient Descent Updates

The update equations resemble those of regularized linear regression, but use logistic regression predictions.

For each weight \(w_j\):

\[
w_j := w_j-\alpha\left[
\frac{1}{m}\sum_{i=1}^{m}
\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})-y^{(i)}\right)x_j^{(i)}
+\frac{\lambda}{m}w_j
\right]
\]

For the bias \(b\):

\[
b := b-\alpha\frac{1}{m}\sum_{i=1}^{m}
\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})-y^{(i)}\right)
\]

The difference from linear regression lies in the prediction function and the underlying cost function. The regularization contribution to the weight update has the same form.

## 3. Understanding the Regularization Parameter

The parameter \(\lambda\) determines the strength of the regularization penalty.

| Regularization | Effect |
|---|---|
| \(\lambda=0\) | No regularization; overfitting may occur. |
| Appropriate \(\lambda\) | Balances fitting the training data with controlling model complexity. |
| Very large \(\lambda\) | Forces weights toward zero and may cause underfitting. |

Choosing an appropriate value helps balance model complexity and predictive performance. In practice, \(\lambda\) is typically selected using validation data or cross-validation.

## 4. Practical Considerations

- **Weights versus bias:** Regularization is typically applied to the feature weights, excluding the bias term.
- **Many features:** Regularization is particularly useful when a model has many input features relative to the available training data.
- **Polynomial features:** It can control excessively complex regression curves and classification boundaries without requiring manual removal of every additional feature.
- **Gradient descent:** The regularization term must be included in the weight updates, not in the bias update.
- **Model evaluation:** Training performance alone is insufficient. Validation or test data should be used to assess generalization.
- **Learning rate:** The learning rate controls the size of each update, while \(\lambda\) controls the strength of the penalty. They serve different purposes.

## 5. Linear vs. Logistic Regression Regularization

| Aspect | Linear Regression | Logistic Regression |
|---|---|---|
| Task | Predict continuous values | Predict class probabilities |
| Prediction function | Linear function | Sigmoid function applied to a linear combination |
| Base cost | Squared error | Binary cross-entropy |
| Regularization penalty | Squared weights | Squared weights |
| Weight updates | Standard gradient plus penalty | Standard gradient plus penalty |
| Bias update | Not regularized | Not regularized |
| Main objective | Reduce overfitting in numerical predictions | Reduce overfitting in classification boundaries |

## Key Takeaways

- Regularization reduces overfitting by penalizing large model weights.
- It modifies the cost function by adding a penalty proportional to the sum of squared weights.
- Gradient descent incorporates an additional term, \(\frac{\lambda}{m}w_j\), into each weight update.
- The bias term is updated as usual and is typically excluded from regularization.
- Weight shrinkage discourages overly complex models and can improve generalization.
- The same regularization principle applies to both linear and logistic regression, even though their prediction functions and base cost functions differ.
- The next step in machine learning is to extend these concepts to neural networks, which also rely on cost functions, gradient descent, and parameter optimization.
