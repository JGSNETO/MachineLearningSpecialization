# Gradient Descent

## Overview

Gradient descent is an optimization algorithm used in machine learning to find parameter values that minimize a **cost function**.

It is fundamental to linear regression and also applies to more complex models with many parameters, including neural networks.

The core idea is simple:

> Start with parameter guesses and repeatedly adjust them in the direction that reduces the cost.

---

## 1. The Cost Function

A machine learning model has parameters that determine its predictions.

For a simple linear regression model:

```text
f(x) = wx + b
```

- `w` — model parameter controlling the slope
- `b` — model parameter controlling the intercept

The **cost function** measures how well the model's predictions match the training data.

The goal of gradient descent is to find values of `w` and `b` that produce a small cost.

---

## 2. Gradient Descent Intuition

Imagine the cost function as a landscape containing hills and valleys.

- High points represent high cost.
- Low points represent low cost.
- A valley represents a minimum of the cost function.

Gradient descent behaves like walking downhill.

At each step, it determines the direction in which the cost decreases most rapidly and moves a small distance in that direction.

The process is repeated until the algorithm reaches or approaches a minimum.

---

## 3. The Parameter Update Rule

The fundamental gradient descent update rule is:

```text
new parameter = old parameter - learning rate × derivative
```

For a parameter such as `w`:

```text
w = w - α × ∂J/∂w
```

For `b`:

```text
b = b - α × ∂J/∂b
```

Where:

- `w` and `b` are model parameters.
- `J` is the cost function.
- `α` (alpha) is the **learning rate**.
- `∂J/∂w` is the derivative of the cost with respect to `w`.
- `∂J/∂b` is the derivative of the cost with respect to `b`.

The equal sign in these expressions represents **assignment**: the newly calculated value replaces the old parameter value.

---

## 4. The Learning Rate

The learning rate, usually represented by `α`, determines how large each update step is.

A small learning rate produces small steps:

```text
small α → small updates
```

A larger learning rate produces larger steps:

```text
large α → large updates
```

The learning rate therefore controls how quickly gradient descent moves across the cost function.

If the steps are too small, optimization can take a long time.

If the steps are too large, the algorithm can potentially overshoot the minimum and fail to converge properly.

---

## 5. Understanding the Derivative

The derivative tells us the **slope of the cost function** at the current parameter value.

It provides two important pieces of information:

1. **Direction** — whether the cost increases or decreases as the parameter changes.
2. **Magnitude** — how steep the cost function is at that point.

Conceptually:

```text
positive derivative → move parameter downward
negative derivative → move parameter upward
```

This happens because the update subtracts the derivative:

```text
parameter = parameter - α × derivative
```

### Positive derivative

If the derivative is positive:

```text
parameter - positive value
```

The parameter decreases.

### Negative derivative

If the derivative is negative:

```text
parameter - negative value
```

The parameter increases.

In both cases, the update attempts to move toward a lower cost.

---

## 6. Why the Derivative Guides the Direction

Consider a point on a cost curve.

If the curve has a positive slope, moving toward smaller parameter values can reduce the cost.

If the curve has a negative slope, moving toward larger parameter values can reduce the cost.

Therefore, the derivative gives gradient descent the information it needs to determine which direction to move.

The learning rate determines how far it moves.

Together:

```text
derivative → direction and steepness
learning rate → step size
```

---

## 7. Iterative Optimization

Gradient descent does not normally find the optimal parameters in a single calculation.

Instead, it repeatedly performs the following process:

1. Start with initial parameter values.
2. Calculate the cost.
3. Calculate the derivatives.
4. Update the parameters.
5. Recalculate the cost.
6. Repeat.

Conceptually:

```text
Initial parameters
        ↓
Calculate derivatives
        ↓
Update parameters
        ↓
Cost decreases
        ↓
Calculate derivatives again
        ↓
Update parameters again
        ↓
Repeat until convergence
```

Initial parameter values are often chosen as zero, although other starting values can also be used.

---

## 8. Simultaneous Parameter Updates

When a model has multiple parameters, all parameter updates should be calculated using the **same previous parameter values**.

For example, with `w` and `b`:

```text
w_new = w_old - α × ∂J/∂w
b_new = b_old - α × ∂J/∂b
```

Both new values should be calculated first.

Only after calculating both values should they replace the old values.

Conceptually:

```text
Calculate w_new
Calculate b_new
        ↓
Assign w = w_new
Assign b = b_new
```

Updating one parameter and immediately using that new value to calculate the next parameter update changes the algorithm and is not the intended simultaneous gradient descent update.

---

## 9. Local Minima and Starting Points

A cost function can potentially contain multiple valleys, or **local minima**.

Gradient descent follows the slope from its current starting point.

Depending on the shape of the cost function and the starting point, the algorithm can converge to different local minima.

Once gradient descent reaches a local minimum, the derivative is approximately zero, so there is no longer a strong downhill direction to follow.

For simple convex cost functions, such as the cost function commonly used for linear regression, the situation is simpler because there is a single global minimum.

---

## 10. Convergence

Gradient descent continues updating parameters until the algorithm reaches a point where further updates produce little or no meaningful improvement.

This is called **convergence**.

Near a minimum:

```text
derivative ≈ 0
```

Therefore:

```text
parameter update ≈ 0
```

The parameters become stable because the algorithm has reached or approached a minimum of the cost function.

---

## 11. Gradient Descent in Linear Regression

For linear regression with parameters `w` and `b`, gradient descent repeatedly applies:

```text
w = w - α × ∂J/∂w
b = b - α × ∂J/∂b
```

The derivatives tell the algorithm how the cost changes when `w` or `b` changes.

The learning rate controls the size of the movement.

The repeated updates gradually move the model toward parameter values that minimize the cost function.

---

## 12. General Gradient Descent Pattern

The same principle applies when a model has many parameters.

For a parameter `θ`:

```text
θ = θ - α × ∂J/∂θ
```

The important idea is that gradient descent does not depend on having only two parameters.

A neural network, for example, can contain thousands or millions of parameters, but the fundamental optimization principle remains the same:

```text
Calculate gradient
        ↓
Determine downhill direction
        ↓
Take a step
        ↓
Repeat
```

---

## 13. Key Concepts

| Concept | Meaning |
|---|---|
| Cost function | Measures how well the model performs |
| Parameter | A value the model learns, such as `w` or `b` |
| Derivative | Indicates the slope of the cost function |
| Gradient | Collection of derivatives for multiple parameters |
| Learning rate | Controls the size of each update |
| Gradient descent | Iterative algorithm for minimizing the cost |
| Local minimum | A point lower than its immediate surroundings |
| Convergence | When parameter updates become very small |

---

## 14. Core Intuition

The entire algorithm can be remembered with three ideas:

**Derivative**

> Which direction should I move?

**Learning rate**

> How far should I move?

**Gradient descent**

> Keep moving in the direction that reduces the cost until reaching a minimum.

The algorithm can therefore be viewed as repeatedly making small downhill steps on the cost-function landscape.

---

## 15. Implementation Example

A simple implementation for a model with parameters `w` and `b` can be written as:

```python
import numpy as np

x_train = np.array([1, 2, 3, 4, 5], dtype=float)
y_train = np.array([2, 4, 6, 8, 10], dtype=float)

w = 0.0
b = 0.0
alpha = 0.01
iterations = 1000

m = len(x_train)

for _ in range(iterations):
    predictions = w * x_train + b
    errors = predictions - y_train

    dj_dw = (1 / m) * np.sum(errors * x_train)
    dj_db = (1 / m) * np.sum(errors)

    w_new = w - alpha * dj_dw
    b_new = b - alpha * dj_db

    w = w_new
    b = b_new

print(f"w = {w:.4f}")
print(f"b = {b:.4f}")
```

This implementation demonstrates the complete optimization cycle:

1. Generate predictions.
2. Calculate prediction errors.
3. Calculate the derivatives.
4. Calculate new values for both parameters.
5. Update both parameters simultaneously.
6. Repeat the process.

---

## 16. Important Takeaways

- Gradient descent minimizes a cost function by iteratively updating model parameters.
- The derivative determines the direction and steepness of the update.
- The learning rate determines the size of the update.
- A positive derivative causes the parameter to decrease.
- A negative derivative causes the parameter to increase.
- Multiple parameters should be updated simultaneously.
- Gradient descent can converge to different local minima depending on the cost landscape and starting point.
- Near a minimum, the derivative approaches zero and parameter updates become very small.
- Gradient descent is not limited to linear regression; the same optimization principle is widely used in machine learning and neural networks.
- Detailed calculus knowledge is not required to understand the basic intuition or implementation of gradient descent.
- Choosing an appropriate learning rate is critical for effective optimization.

````python
'''
Basic Gradient Descent

Goal:
    Minimize the cost function:

        J(w) = (w - 3)^2

The minimum occurs when:
    w = 3

Gradient descent starts with an initial guess
and repeatedly moves w toward the minimum.
'''

w = 0.0
alpha = 0.1

for _ in range(20):

    '''
    Calculate the derivative of the cost function.

    For:
        J(w) = (w - 3)^2

    The derivative is:
        dJ/dw = 2(w - 3)
    '''

    derivative = 2 * (w - 3)

    '''
    Gradient descent update rule:

        new parameter = old parameter - learning rate × derivative
    '''

    w = w - alpha * derivative

print(f"Final w = {w:.4f}")
````

````python
'''
Gradient Descent with Cost Tracking

This example shows how the cost decreases
as gradient descent updates the parameter.
'''

w = 0.0
alpha = 0.1
iterations = 20

for i in range(iterations):

    '''
    Calculate the current cost.

        J(w) = (w - 3)^2
    '''

    cost = (w - 3) ** 2

    '''
    Calculate the derivative.

        dJ/dw = 2(w - 3)
    '''

    derivative = 2 * (w - 3)

    '''
    Update the parameter.

        w = w - alpha × derivative
    '''

    w = w - alpha * derivative

    print(
        f"Iteration {i + 1}: "
        f"w = {w:.4f}, "
        f"cost = {cost:.4f}"
    )
````

````python
'''
Positive vs. Negative Derivative

The derivative determines the direction
of the gradient descent step.

Positive derivative:
    The parameter decreases.

Negative derivative:
    The parameter increases.
'''

alpha = 0.1

'''
Start on the right side of the minimum.

Here:
    w = 4

The derivative is positive, so w decreases.
'''

w = 4.0

derivative = 2 * (w - 3)

w_new = w - alpha * derivative

print(f"Positive derivative: {derivative:.2f}")
print(f"w moves from {w:.2f} to {w_new:.2f}")

'''
Start on the left side of the minimum.

Here:
    w = 2

The derivative is negative, so w increases.
'''

w = 2.0

derivative = 2 * (w - 3)

w_new = w - alpha * derivative

print(f"Negative derivative: {derivative:.2f}")
print(f"w moves from {w:.2f} to {w_new:.2f}")
````

````python
'''
Simultaneous Parameter Updates

A model can have multiple parameters.

For linear regression:

    f(x) = wx + b

Both w and b must be updated using
the values from the same iteration.

The new values are calculated first:

    w_new = w - alpha × dJ/dw
    b_new = b - alpha × dJ/db

Only after both calculations are complete
are the original parameters replaced.
'''

w = 0.0
b = 0.0
alpha = 0.1

'''
Example derivatives.

These would normally be calculated
from the training data and cost function.
'''

dj_dw = -4.0
dj_db = -2.0

'''
Calculate both new parameter values
before changing w or b.
'''

w_new = w - alpha * dj_dw
b_new = b - alpha * dj_db

'''
Apply both updates simultaneously.
'''

w = w_new
b = b_new

print(f"w = {w}")
print(f"b = {b}")
````

````python
'''
Complete Gradient Descent Implementation

This example implements gradient descent
for a simple linear regression model.

Model:

    f(x) = wx + b

Training data:

    x = [1, 2, 3, 4, 5]
    y = [2, 4, 6, 8, 10]

The algorithm learns values approximately:

    w = 2
    b = 0

The implementation includes:

    1. Predictions
    2. Errors
    3. Cost calculation
    4. Derivatives
    5. Parameter updates
    6. Iterative optimization
'''

import numpy as np


'''
Training data.

x_train contains the input features.
y_train contains the target values.
'''

x_train = np.array([1, 2, 3, 4, 5], dtype=float)
y_train = np.array([2, 4, 6, 8, 10], dtype=float)


'''
Initialize the model parameters.

w = weight
b = bias
'''

w = 0.0
b = 0.0


'''
Learning rate.

This controls the size of each gradient
descent step.
'''

alpha = 0.01


'''
Number of gradient descent iterations.
'''

iterations = 1000


'''
Number of training examples.
'''

m = len(x_train)


'''
Gradient descent loop.

Each iteration updates w and b
to reduce the cost.
'''

for i in range(iterations):

    '''
    Calculate predictions using:

        f(x) = wx + b
    '''

    predictions = w * x_train + b

    '''
    Calculate prediction errors.

        error = prediction - actual value
    '''

    errors = predictions - y_train

    '''
    Calculate the cost.

        J(w,b) = 1/(2m) × Σ(error²)

    The cost measures how far the predictions
    are from the actual training values.
    '''

    cost = (1 / (2 * m)) * np.sum(errors ** 2)

    '''
    Calculate the derivative of the cost
    with respect to w.

        dJ/dw = 1/m × Σ(error × x)

    This determines how w should change.
    '''

    dj_dw = (1 / m) * np.sum(errors * x_train)

    '''
    Calculate the derivative of the cost
    with respect to b.

        dJ/db = 1/m × Σ(error)

    This determines how b should change.
    '''

    dj_db = (1 / m) * np.sum(errors)

    '''
    Calculate the new parameter values.

    IMPORTANT:

    Both new values are calculated before
    replacing the old values.

    This implements simultaneous updates.
    '''

    w_new = w - alpha * dj_dw
    b_new = b - alpha * dj_db

    '''
    Apply the parameter updates.
    '''

    w = w_new
    b = b_new


'''
Display the learned parameters and final cost.
'''

print(f"w = {w:.4f}")
print(f"b = {b:.4f}")
print(f"Final cost = {cost:.4f}")
````

````python
'''
Effect of the Learning Rate

The learning rate controls the size
of each gradient descent step.

Small learning rate:
    Smaller steps
    Slower convergence

Larger learning rate:
    Larger steps
    Potentially faster convergence

Very large learning rate:
    Can overshoot the minimum
    and potentially fail to converge
'''


def gradient_descent(alpha, iterations=20):

    '''
    Start with an initial parameter value.
    '''

    w = 0.0

    '''
    Perform gradient descent.
    '''

    for _ in range(iterations):

        '''
        Calculate the derivative.

            dJ/dw = 2(w - 3)
        '''

        derivative = 2 * (w - 3)

        '''
        Update w.

            w = w - alpha × derivative
        '''

        w = w - alpha * derivative

    return w


'''
Test several different learning rates.
'''

learning_rates = [0.01, 0.1, 0.5, 1.1]

for alpha in learning_rates:

    w = gradient_descent(alpha)

    print(
        f"alpha = {alpha}: "
        f"w = {w:.4f}"
    )
````
