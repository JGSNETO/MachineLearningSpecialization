# Neural Network Layers and Forward Propagation

## Overview

This section explores how neural networks compute predictions by stacking layers of neurons and passing information sequentially through the network.

The main topics are:

- Computation performed by individual neurons and layers
- Neural network layer notation
- Mathematical formulation of activations
- Forward propagation for inference
- A handwritten digit recognition example
- The distinction between inference and training

## 1. Computation Within a Layer

A neural network layer consists of multiple neurons that process the same input vector. Each neuron has its own weights and bias, allowing it to learn a different transformation of the input.

For a layer with four input features and three neurons:

- The input vector contains four values.
- Each neuron calculates a weighted sum of the inputs and adds a bias.
- An activation function transforms the result.
- The layer produces three activation values, one per neuron.

The resulting activation vector becomes the input to the next layer.

```text
Input vector: 4 features
        |
        v
+----------------------+
|      Layer 1         |
|                      |
|  Neuron 1 --> a₁⁽¹⁾  |
|  Neuron 2 --> a₂⁽¹⁾  |
|  Neuron 3 --> a₃⁽¹⁾  |
+----------------------+
        |
        v
Activation vector: 3 values
```

Each neuron has independent parameters, even when neurons belong to the same layer.

## 2. Neural Network Notation

Consistent mathematical notation makes it easier to describe networks of different sizes and depths.

### Layer numbering

The input is represented as Layer 0. Hidden and output layers follow sequentially.

For example, a network with three hidden layers and one output layer has the following structure:

```text
Input        Hidden       Hidden       Hidden       Output
Layer 0  --> Layer 1  --> Layer 2  --> Layer 3  --> Layer 4
```

The network has four computational layers when counting hidden and output layers, excluding the input layer.

### Activations and parameters

The notation uses superscripts to identify the layer and subscripts to identify the neuron.

| Symbol | Meaning |
|---|---|
| \(a_j^{[l]}\) | Activation of neuron \(j\) in layer \(l\) |
| \(w_j^{[l]}\) | Weight vector associated with neuron \(j\) in layer \(l\) |
| \(b_j^{[l]}\) | Bias associated with neuron \(j\) in layer \(l\) |
| \(a^{[l]}\) | Activation vector produced by layer \(l\) |
| \(x\) or \(a^{[0]}\) | Original input feature vector |
| \(g\) | Activation function |

For example, \(a_2^{[3]}\) represents the activation of the second neuron in the third layer.

Representing the input as \(a^{[0]}=x\) allows every layer to follow the same mathematical convention.

## 3. Mathematical Formulation of a Neuron

The activation of neuron \(j\) in layer \(l\) is calculated using the activations from the previous layer:

\[
a_j^{[l]} =
g\left(
(w_j^{[l]})^T a^{[l-1]}
+b_j^{[l]}
\right)
\]

The computation has three main steps:

1. **Weighted sum:** Multiply the previous layer's activations by the neuron's weights and sum the results.
2. **Bias addition:** Add the neuron's bias.
3. **Activation:** Apply the activation function \(g\).

In this example, the activation function is the sigmoid function:

\[
g(z)=\frac{1}{1+e^{-z}}
\]

The sigmoid function maps any real-valued input to a value between 0 and 1, making it useful for binary classification outputs.

Other activation functions, such as ReLU, can be used in hidden layers depending on the architecture and learning task.

### Computing an entire layer

The same computation is performed independently for every neuron in a layer. Their outputs are collected into an activation vector:

\[
a^{[l]} =
\begin{bmatrix}
a_1^{[l]}\\
a_2^{[l]}\\
\vdots\\
a_{n_l}^{[l]}
\end{bmatrix}
\]

Here, \(n_l\) is the number of neurons in layer \(l\).

This vectorized representation provides the foundation for implementing neural networks efficiently using numerical libraries and machine learning frameworks.

## 4. Forward Propagation

Forward propagation is the algorithm used to compute a neural network's prediction from an input.

The input is passed through each layer in order. Every layer calculates its activations and supplies them to the next layer until the output is produced.

For a network with three computational layers:

\[
a^{[1]}=g_1(W^{[1]}x+b^{[1]})
\]

\[
a^{[2]}=g_2(W^{[2]}a^{[1]}+b^{[2]})
\]

\[
a^{[3]}=g_3(W^{[3]}a^{[2]}+b^{[3]})
\]

The final activation \(a^{[3]}\) represents the network's output.

The matrices \(W^{[l]}\) and vectors \(b^{[l]}\) contain the weights and biases for the corresponding layers. The function \(g_l\) represents the activation function used in each layer.

```text
Input x = a⁽⁰⁾
      |
      v
Layer 1: calculate a⁽¹⁾
      |
      v
Layer 2: calculate a⁽²⁾
      |
      v
Layer 3: calculate a⁽³⁾
      |
      v
Final prediction
```

The defining characteristic of forward propagation is that information flows from the input toward the output. The model uses its current parameters to calculate a prediction without updating those parameters.

## 5. Example: Handwritten Digit Recognition

Consider a neural network designed to distinguish between handwritten zeros and ones.

### Input representation

The example uses an image containing \(8 \times 8\) pixels.

Each pixel contributes one input feature, so the flattened image contains:

\[
8 \times 8 = 64 \text{ features}
\]

The input vector represents the pixel intensities of the image.

### Network architecture

The example network contains two hidden layers and one output neuron.

| Layer | Number of neurons | Purpose |
|---|---:|---|
| Input | 64 features | Represents pixel intensities |
| Hidden Layer 1 | 25 | Learns intermediate representations |
| Hidden Layer 2 | 15 | Learns further representations |
| Output Layer | 1 | Estimates the probability of digit one |

The forward propagation sequence is:

```text
64 pixel features
       |
       v
Hidden Layer 1
    25 neurons
       |
       v
Hidden Layer 2
    15 neurons
       |
       v
Output Layer
     1 neuron
       |
       v
P(digit = 1 | image)
```

The output can be interpreted as the estimated probability that the image represents the digit one.

For example, an output of 0.82 indicates an estimated probability of 82% for digit one.

### Converting the probability into a class

For binary classification, a threshold can convert the probability into a discrete prediction.

Using a threshold of 0.5:

\[
\hat{y}=
\begin{cases}
1 & \text{if } a^{[3]}\geq 0.5\\
0 & \text{if } a^{[3]}<0.5
\end{cases}
\]

This produces a class label of either zero or one. The threshold is a configurable decision rule, not an inherent requirement of every neural network.

## 6. Forward Propagation in Practice

The algorithm follows a simple sequence:

1. Receive the input feature vector.
2. Compute the activations of the first layer.
3. Pass those activations to the second layer.
4. Repeat for every subsequent layer.
5. Return the final output.
6. If required, apply a decision threshold to obtain a class label.

Frameworks such as TensorFlow perform these operations efficiently using tensor operations and optimized numerical computation.

In practice, the calculations are typically vectorized rather than implemented as individual scalar operations for every neuron.

## 7. Forward Propagation vs. Backpropagation

Forward propagation and backpropagation have different purposes.

| Forward propagation | Backpropagation |
|---|---|
| Computes predictions | Computes gradients of the loss |
| Passes activations from input to output | Propagates gradient information backward through the network |
| Used during inference | Used during training to determine parameter updates |
| Uses the current weights and biases | Helps an optimizer adjust the weights and biases |

During training, forward propagation calculates predictions, a loss function measures prediction error, and backpropagation calculates gradients. An optimizer then uses those gradients to update the model parameters.

During ordinary inference, the trained parameters remain fixed and forward propagation is sufficient to generate predictions.

## Key Takeaways

- Each layer transforms an input activation vector into a new activation vector.
- Superscripts identify layers, while subscripts identify individual neurons.
- Every neuron uses weights, a bias, and an activation function to compute its output.
- Forward propagation applies these computations sequentially from input to output.
- The final output can represent a probability and can be thresholded for binary classification.
- Backpropagation is used during training to calculate gradients; forward propagation is used both during training and inference.

Understanding these computations provides the mathematical foundation for implementing, debugging, and optimizing neural networks in frameworks such as TensorFlow.
