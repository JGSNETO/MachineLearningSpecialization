# Training a Neural Network Using TensorFlow

## 1. Introduction

Training a neural network consists of finding the optimal parameters (weights and biases) that allow a model to make accurate predictions.

The training process follows the same fundamental principles as **logistic regression**:

1. Define how the model computes predictions.
2. Measure prediction errors using a loss function.
3. Minimize the cost function by updating model parameters.

TensorFlow automates most of the mathematical operations involved in this process, including forward propagation, gradient calculation, backpropagation, and parameter optimization.

This example uses a neural network for handwritten digit recognition.

Before training, we must define the network architecture.

The architecture determines:
- The number of layers.

## 3. The Three Steps of Neural Network Training

Neural network training can be divided into three fundamental steps.

### Step 1: Compute the model output (Forward Propagation)

The first step is defining how the neural network transforms input features into predictions.

Each neuron receives inputs, multiplies them by weights, adds a bias, and applies an activation function.

The computation can be represented as:

\[
z = w_1x_1+w_2x_2+\cdots+w_nx_n+b
\]

\[
a = g(z)
\]

Where:

- \(x\): Input features.
- \(w\): Model weights.
- \(b\): Bias.
- \(z\): Weighted sum of inputs.
- \(g\): Activation function.
- \(a\): Neuron output.

The output of one layer becomes the input of the next layer.

This process is called **forward propagation**.

### Step 2: Compute the loss and cost

After generating a prediction, the model needs to evaluate how accurate that prediction is.

This is accomplished using a loss function.

A loss function measures the difference between the predicted output and the actual target.

For binary classification, a common choice is Binary Cross-Entropy:

\[
L(\hat{y},y) =
-\left[y\log(\hat{y})+(1-y)\log(1-\hat{y})\right]
\]

Where:

- \(y\): Actual label (0 or 1).
- \(\hat{y}\): Predicted probability.

The cost function represents the average loss across all training examples:

\[
J(w,b)=\frac{1}{m}\sum_{i=1}^{m}L(\hat{y}^{(i)},y^{(i)})
\]

Where:

- \(m\): Number of training examples.
- \(L\): Loss for an individual example.
- \(J\): Overall cost.

**Loss vs. Cost**

| Concept | Meaning |
|---|---|
| Loss | Error for one training example |
| Cost | Average loss across the training dataset |

The objective is to minimize the cost function.

### Step 3: Minimize the cost function (Optimization)

Once the cost is calculated, the model needs to adjust its parameters to improve predictions.

This is accomplished using optimization algorithms such as Gradient Descent.

Gradient descent updates the weights and biases in the direction that reduces the cost.

\[
w := w-\alpha\frac{\partial J}{\partial w}
\]

\[
b := b-\alpha\frac{\partial J}{\partial b}
\]

Where:

- \(\alpha\): Learning rate.
- \(\frac{\partial J}{\partial w}\): Gradient of the cost with respect to the weights.
- \(\frac{\partial J}{\partial b}\): Gradient of the cost with respect to the bias.

The gradients indicate how the cost changes when the parameters change.

**Backpropagation** is the algorithm used to calculate these gradients by applying the chain rule from the output layer back through the network.

TensorFlow automatically calculates the gradients and performs the parameter updates during training.

## 4. Implementing the Neural Network Using TensorFlow

Instead of implementing forward propagation, backpropagation, and gradient descent manually, we can use TensorFlow.

### 4.1 Import TensorFlow

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

### 4.2 Define the neural network

```python
model = Sequential([
    tf.keras.Input(shape=(784,)),
    Dense(25, activation='relu'),
    Dense(15, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

Explanation:

- `Sequential`: Creates a neural network where layers are connected sequentially.
- `Input(shape=(784,))`: Defines the input features.
- `Dense(25)`: Creates a fully connected hidden layer with 25 neurons.
- `Dense(15)`: Creates another fully connected hidden layer with 15 neurons.
- `Dense(1)`: Creates the output layer with one neuron.
- `relu`: Activation function used in the hidden layers.
- `sigmoid`: Produces an output between 0 and 1, suitable for binary classification.

### 4.3 Compile the model

```python
model.compile(
    loss='binary_crossentropy',
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001)
)
```

Compilation configures the training process.

**Loss function:**
- `binary_crossentropy`: Measures prediction error for binary classification.

**Optimizer:**
- `Adam`: Optimization algorithm that updates the model parameters based on calculated gradients.
- `learning_rate`: Controls the size of parameter updates.

Other commonly used loss functions include:

| Task | Loss function |
|---|---|
| Binary classification | Binary Cross-Entropy |
| Multiclass classification | Categorical Cross-Entropy |
| Regression | Mean Squared Error |

Parameters:

- `X_train`: Training input features.

During training, TensorFlow automatically performs:

1. Forward propagation.
2. Loss calculation.
3. Backpropagation and gradient calculation.
4. Parameter updates using the optimizer.

This process repeats until the specified number of epochs is completed.

---

## 5. Understanding Epochs

An epoch represents one complete pass through the entire training dataset.

For example, if the training dataset contains 1,000 images:

- 1 epoch means the model has processed all 1,000 images once.
- 10 epochs means the model has processed the dataset ten times.

During each epoch, the model updates its parameters to reduce prediction errors.

**Note:** An epoch is not necessarily one parameter update. When training uses mini-batches, the optimizer updates parameters multiple times within each epoch.

---

## 6. TensorFlow vs. Manual Implementation

Understanding the underlying mathematics is important, even when using a high-level library.

| Training operation | Manual implementation | TensorFlow |
|---|---|---|
| Forward propagation | Implement equations | Automatically calculated |
| Loss calculation | Implement loss function | Built-in loss functions |
| Gradient calculation | Implement backpropagation | Automatic differentiation |
| Parameter updates | Implement gradient descent | Built-in optimizers |
| Training loop | Implement manually | `model.fit()` |

Modern deep learning development typically relies on mature frameworks such as TensorFlow and PyTorch.

However, understanding the underlying mechanics makes it easier to troubleshoot models, select appropriate configurations, and improve performance.

---

## 7. Complete Training Workflow

The overall training process can be summarized as:

1. **Define the architecture:** Configure layers, neurons, and activation functions.
2. **Compile the model:** Specify the loss function and optimizer.
3. **Train the model:** Call `model.fit()` with the training dataset.
4. **Evaluate performance:** Check how well the trained model performs on unseen data.

The central objective is to find weights and biases that minimize the cost function and improve the model's predictions.

---

## 8. Key Takeaways

- Neural network training follows the same fundamental principles as logistic regression.
- Forward propagation calculates predictions from input features.
- The loss function measures prediction errors for individual examples.
- The cost function calculates the average loss across the training dataset.
- Backpropagation calculates gradients used to update model parameters.
- Gradient descent and other optimizers minimize the cost function.
- TensorFlow automates these operations through its training framework.
- `model.fit()` executes the training process over a specified number of epochs.

**Main concept to remember:**

A neural network learns by repeatedly making predictions, measuring errors, calculating gradients, and updating its weights and biases to reduce those errors.
