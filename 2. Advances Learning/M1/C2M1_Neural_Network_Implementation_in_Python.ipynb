# Forward Propagation in Neural Networks: From Scratch with Python

## 1. Introduction

Forward propagation is the process through which a neural network receives input data and passes it through its layers to produce an output.

Each layer transforms the activations received from the previous layer using weights, biases, and an activation function.

Understanding how forward propagation works is essential for understanding the internal mechanics of neural networks and how high-level machine learning frameworks such as TensorFlow and PyTorch perform their computations.

This tutorial explores forward propagation from scratch using Python, progressing from individual neuron calculations to a generalized implementation capable of handling multiple dense layers.

---

# Part I — Theory

## 2. The Computation of a Neuron

A neuron receives input values, multiplies each input by its corresponding weight, adds a bias, and applies an activation function.

The computation has two main steps.

### 2.1 Weighted Sum

The first step is calculating the weighted sum of the inputs and adding the bias.

\[
z = \mathbf{w}\cdot\mathbf{x}+b
\]

Where:

- \(x\): input feature vector.
- \(w\): weight vector.
- \(b\): bias.
- \(z\): weighted sum before activation.

The dot product is calculated as:

\[
z = w_1x_1+w_2x_2+\cdots+w_nx_n+b
\]

Each weight determines how strongly its corresponding input contributes to the neuron's computation.

### 2.2 Activation Function

After calculating the weighted sum, the neuron applies an activation function.

For this implementation, we use the sigmoid function:

\[
g(z)=\frac{1}{1+e^{-z}}
\]

The neuron's final activation is:

\[
a=g(z)
\]

The sigmoid function transforms the weighted sum into a value between 0 and 1.

The activation becomes the output of the neuron and can be passed to the next layer.

---

## 3. From Individual Neurons to Dense Layers

A neural network layer typically contains multiple neurons.

A dense layer is a layer in which every neuron receives input from every activation in the previous layer.

Instead of representing each neuron's weights independently, we can organize all weights into a matrix.

Consider a layer containing three neurons, each receiving two input features.

The weight matrix is:

\[
W=
\begin{bmatrix}
w_{11} & w_{12}\\
w_{21} & w_{22}\\
w_{31} & w_{32}
\end{bmatrix}
\]

The bias values are organized into a vector:

\[
\mathbf{b}=
\begin{bmatrix}
b_1\\
b_2\\
b_3
\end{bmatrix}
\]

The input vector is:

\[
\mathbf{x}=
\begin{bmatrix}
x_1\\
x_2
\end{bmatrix}
\]

The weighted sums for all neurons can now be calculated using matrix multiplication:

\[
\mathbf{z}=W\mathbf{x}+\mathbf{b}
\]

The activation function is applied to every element of the resulting vector:

\[
\mathbf{a}=g(\mathbf{z})
\]

This produces the output activations of the entire layer.

### 3.1 Understanding the Dimensions

For a layer with:

- \(n\) input features.
- \(m\) neurons.

The dimensions are:

| Parameter | Dimensions |
|---|---|
| Input vector \(x\) | \(n \times 1\) |
| Weight matrix \(W\) | \(m \times n\) |
| Bias vector \(b\) | \(m \times 1\) |
| Weighted sum \(z\) | \(m \times 1\) |
| Activation vector \(a\) | \(m \times 1\) |

The number of rows in the weight matrix corresponds to the number of neurons, while the number of columns corresponds to the number of input features.

This organization makes it possible to calculate the activations of an entire layer using matrix operations rather than manually computing each neuron.

---

## 4. Forward Propagation Across Multiple Layers

A neural network can contain multiple dense layers connected sequentially.

The output activations of one layer become the input activations of the next layer.

For a network with two layers:

\[
\mathbf{a}^{[1]}=g(W^{[1]}\mathbf{x}+\mathbf{b}^{[1]})
\]

\[
\mathbf{a}^{[2]}=g(W^{[2]}\mathbf{a}^{[1]}+\mathbf{b}^{[2]})
\]

Where:

- \(x\) is the original input.
- \(a^{[1]}\) is the activation vector from the first layer.
- \(a^{[2]}\) is the activation vector from the second layer.
- \(W^{[l]}\) and \(b^{[l]}\) represent the weights and biases of layer \(l\).

The same computation is repeated for every layer until the network produces its final output.

### 4.1 The General Forward Propagation Formula

For a network containing \(L\) layers:

\[
\mathbf{a}^{[0]}=\mathbf{x}
\]

For each layer \(l\):

\[
\mathbf{z}^{[l]}=W^{[l]}\mathbf{a}^{[l-1]}+\mathbf{b}^{[l]}
\]

\[
\mathbf{a}^{[l]}=g(\mathbf{z}^{[l]})
\]

The final prediction is:

\[
\hat{\mathbf{y}}=\mathbf{a}^{[L]}
\]

This formulation provides a general mathematical description of forward propagation, independent of the number of layers or neurons.

---

## 5. Why Generalized Forward Propagation Matters

A neural network implementation should not require manually writing the calculation for every neuron or layer.

By organizing weights and biases into matrices and vectors, we can create a reusable dense-layer function.

This function can be called repeatedly to process any number of layers, provided their dimensions are compatible.

This approach provides several advantages:

- **Scalability:** supports networks with different numbers of neurons and layers.
- **Reusability:** the same function can process every dense layer.
- **Maintainability:** reduces duplicated computation.
- **Understanding:** makes the relationship between mathematical operations and neural network implementation explicit.
- **Debugging:** allows individual layer calculations to be inspected.

Understanding these operations also makes it easier to understand what high-level frameworks such as TensorFlow and PyTorch are doing internally.

# Part II — Python Implementation

## 6. Implementing the Basic Operations

We begin by implementing the sigmoid activation function.

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

The function accepts a numerical value or NumPy array and applies the sigmoid transformation.

Because NumPy supports element-wise operations, the same function can process an entire activation vector.

---

## 7. Implementing Forward Propagation for Individual Neurons

Consider a simplified coffee-roasting model.

The input features represent characteristics of the roasting process, while the model parameters determine how each feature contributes to the prediction.

For this example, assume three neurons in the first layer.

Each neuron has its own weight vector and bias.

```python
# Input features
x = np.array([200, 17])

# Weights for three neurons
w1 = np.array([1, 2])
w2 = np.array([-1, 1])
w3 = np.array([2, -3])

# Biases
b1 = -100
b2 = 50
b3 = -200
```

We calculate the weighted sum and activation of each neuron:

```python
a1 = sigmoid(np.dot(w1, x) + b1)
a2 = sigmoid(np.dot(w2, x) + b2)
a3 = sigmoid(np.dot(w3, x) + b3)

# Output of the first layer
a = np.array([a1, a2, a3])
```

The resulting vector contains the activations of all three neurons.

This approach works, but manually defining every neuron becomes impractical as the network grows.

The next step is to generalize this calculation.

---

## 8. Implementing a Generalized Dense Layer

A dense function represents an entire neural network layer.

It receives:

- `a_in`: activations from the previous layer.
- `W`: weight matrix containing the weights of all neurons.
- `b`: bias vector containing the bias of each neuron.

The function calculates the activation of every neuron and returns the resulting vector.

```python
def dense(a_in, W, b):
    units = W.shape[0]
    a_out = np.zeros(units)

    for j in range(units):
        w = W[j]
        z = np.dot(w, a_in) + b[j]
        a_out[j] = sigmoid(z)

    return a_out
```

### 8.1 Understanding the Implementation

**Step 1 — Determine the number of neurons**

```python
units = W.shape[0]
```

The number of rows in the weight matrix corresponds to the number of neurons in the layer.

**Step 2 — Initialize the output vector**

```python
a_out = np.zeros(units)
```

An output array is created to store the activation of each neuron.

**Step 3 — Iterate through the neurons**

```python
for j in range(units):
```

The function processes each neuron individually.

**Step 4 — Calculate the weighted sum**

```python
w = W[j]
z = np.dot(w, a_in) + b[j]
```

The corresponding weight vector is selected, its dot product with the input activations is calculated, and the neuron's bias is added.

**Step 5 — Apply the activation function**

```python
a_out[j] = sigmoid(z)
```

The sigmoid function transforms the weighted sum into the neuron's activation.

**Step 6 — Return the layer output**

```python
return a_out
```

The completed activation vector becomes the input for the next layer.

---

## 9. Building a Multi-Layer Neural Network

Once the dense function is implemented, we can construct a neural network by calling it sequentially.

Each layer receives the output of the previous layer.

For example, consider a network with three dense layers:

```python
# Input features
x = np.array([200, 17])

# Layer 1: 3 neurons, 2 input features
W1 = np.array([
    [1, 2],
    [-1, 1],
    [2, -3]
])

b1 = np.array([-100, 50, -200])

# Layer 2: 2 neurons, 3 input features
W2 = np.array([
    [1, -1, 2],
    [-2, 1, 1]
])

b2 = np.array([0, -10])

# Layer 3: 1 neuron, 2 input features
W3 = np.array([
    [1, -2]
])

b3 = np.array([5])
```

We can now perform forward propagation:

```python
a1 = dense(x, W1, b1)
a2 = dense(a1, W2, b2)
a3 = dense(a2, W3, b3)

prediction = a3
```

The data flows through the network in the following sequence:

```text
Input Features
      |
      v
  Dense Layer 1
   (3 neurons)
      |
      v
  Dense Layer 2
   (2 neurons)
      |
      v
  Dense Layer 3
   (1 neuron)
      |
      v
   Prediction
```

Notice that the same `dense()` function is reused for all three layers.

The only differences between calls are the input activations, weight matrices, and bias vectors.

---

## 10. Generalizing the Network with a Loop

Instead of manually calling the dense function for every layer, we can organize the parameters into lists and iterate through them.

```python
# Organize the network parameters
W = [W1, W2, W3]
B = [b1, b2, b3]

# Initialize the network input
a = x

# Forward propagation
for i in range(len(W)):
    a = dense(a, W[i], B[i])

prediction = a
```

This implementation is more flexible because the number of layers is no longer hardcoded into the forward propagation logic.

Adding another layer only requires adding its weight matrix and bias vector to the corresponding lists.

The network can therefore process different architectures using the same forward propagation function.

---

## 11. Key Takeaways

- Forward propagation calculates a neural network's output by processing input data through its layers.
- Each neuron calculates a weighted sum, adds a bias, and applies an activation function.
- Dense layers organize neuron weights into matrices and biases into vectors.
- Matrix operations provide an efficient way to calculate activations across an entire layer.
- A reusable `dense()` function allows the same computation to be applied to different layers.
- Multi-layer networks are constructed by passing each layer's output into the next layer.
- A loop can generalize the forward pass to networks with an arbitrary number of layers.

**The central idea:** Forward propagation is a sequence of mathematical transformations. Once the computation of a single dense layer is understood and implemented, the entire forward pass can be constructed by repeatedly applying that same operation.

````python
import numpy as np

# 1. Activation function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


# 2. Dense layer
def dense(a_in, W, b):
    units = W.shape[0]
    a_out = np.zeros(units)

    for j in range(units):
        w = W[j]
        z = np.dot(w, a_in) + b[j]
        a_out[j] = sigmoid(z)

    return a_out


# 3. Input features
x = np.array([200, 17])


# 4. Neural network parameters

# Layer 1: 3 neurons, 2 inputs
W1 = np.array([
    [1, 2],
    [-1, 1],
    [2, -3]
])
b1 = np.array([-100, 50, -200])

# Layer 2: 2 neurons, 3 inputs
W2 = np.array([
    [1, -1, 2],
    [-2, 1, 1]
])
b2 = np.array([0, -10])

# Layer 3: 1 neuron, 2 inputs
W3 = np.array([
    [1, -2]
])
b3 = np.array([5])


# 5. Explicit forward propagation
a1 = dense(x, W1, b1)
a2 = dense(a1, W2, b2)
a3 = dense(a2, W3, b3)

prediction = a3

print("Layer 1:", a1)
print("Layer 2:", a2)
print("Layer 3:", a3)
print("Prediction:", prediction)


# 6. Generalized forward propagation
W = [W1, W2, W3]
B = [b1, b2, b3]

a = x

for i in range(len(W)):
    a = dense(a, W[i], B[i])

print("Generalized prediction:", a)
````
