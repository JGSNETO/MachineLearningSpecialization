# Classification & Logistic Regression

## 1. Introduction to Classification

**Classification** is a supervised machine learning task in which a model predicts a discrete category rather than a continuous numerical value.

Unlike regression, which predicts values such as house prices or temperatures, classification predicts labels belonging to a predefined set of classes.

### Binary Classification

Binary classification involves two possible classes, commonly represented as:

- `0` — Negative class
- `1` — Positive class

Other examples include:

- Email: spam or not spam
- Medical diagnosis: malignant or benign
- Automotive systems: obstacle detected or not detected
- Quality control: defective or non-defective component

The model learns patterns from labeled training data and uses those patterns to classify new observations.

## 2. Why Linear Regression Is Not Suitable for Classification

Linear regression predicts continuous numerical values using a linear function:

\[
f_{\mathbf{w},b}(\mathbf{x})=\mathbf{w}\cdot\mathbf{x}+b
\]

These predictions are not restricted to the interval between 0 and 1. Therefore, they cannot directly represent valid probabilities for binary classification.

A possible workaround is to apply a threshold:

- If the prediction is at least `0.5`, classify it as `1`.
- Otherwise, classify it as `0`.

However, linear regression is sensitive to the distribution of the training data. Adding observations can change the fitted line and shift the resulting decision boundary, potentially producing poor classifications.

**Logistic regression** is better suited to binary classification because it transforms a linear combination of features into a probability between 0 and 1.

## 3. Logistic Regression

Logistic regression is a supervised learning algorithm commonly used for classification.

It combines two functions:

1. A linear function that combines the input features.
2. A sigmoid function that converts the linear result into a probability.

### 3.1 Linear Component

The model first computes:

\[
z=\mathbf{w}\cdot\mathbf{x}+b
\]

For a model with two input features:

\[
z=w_1x_1+w_2x_2+b
\]

Where:

- \(x_1,x_2\) are input features.
- \(w_1,w_2\) are learned weights.
- \(b\) is the bias.
- \(z\) is the linear combination of the features.

The weights determine how strongly each feature influences the result, while the bias shifts the model's output.

### 3.2 Sigmoid Function

The linear output \(z\) can take any real value. The sigmoid function transforms it into a value strictly between 0 and 1:

\[
g(z)=\frac{1}{1+e^{-z}}
\]

The logistic regression prediction is:

\[
f_{\mathbf{w},b}(\mathbf{x})=g(\mathbf{w}\cdot\mathbf{x}+b)
\]

The sigmoid function has an S-shaped curve.

| Input \(z\) | Sigmoid output \(g(z)\) |
|---:|---:|
| Large negative value | Approaches 0 |
| \(-2\) | Approximately 0.119 |
| \(0\) | 0.5 |
| \(2\) | Approximately 0.881 |
| Large positive value | Approaches 1 |

At \(z=0\), the predicted probability is exactly 0.5. Positive values of \(z\) produce probabilities above 0.5, while negative values produce probabilities below 0.5.

## 4. Interpreting Model Predictions

The output of logistic regression represents the estimated probability that the target belongs to the positive class:

\[
f_{\mathbf{w},b}(\mathbf{x})=P(y=1\mid\mathbf{x};\mathbf{w},b)
\]

For binary classification, the probability of the negative class is:

\[
P(y=0\mid\mathbf{x})=1-P(y=1\mid\mathbf{x})
\]

For example, a model predicting a probability of `0.7` estimates a 70% probability of the positive class and a 30% probability of the negative class.

### Making a Classification

Using the standard threshold of 0.5:

\[
\hat{y}=
\begin{cases}
1 & \text{if } f_{\mathbf{w},b}(\mathbf{x})\geq0.5\\
0 & \text{if } f_{\mathbf{w},b}(\mathbf{x})<0.5
\end{cases}
\]

This converts a continuous probability into a discrete class label.

For example:

| Predicted probability | Predicted class |
|---:|---:|
| 0.15 | 0 |
| 0.42 | 0 |
| 0.50 | 1 |
| 0.73 | 1 |
| 0.96 | 1 |

The threshold does not have to be 0.5 in every application. Depending on the consequences of false positives and false negatives, a different threshold may be more appropriate.

## 5. Decision Boundaries

A **decision boundary** separates regions of the feature space where a classifier predicts different classes.

In logistic regression, using a threshold of 0.5, the decision boundary occurs when:

\[
f_{\mathbf{w},b}(\mathbf{x})=0.5
\]

Because the sigmoid function equals 0.5 when its input is zero, the boundary can be expressed as:

\[
\mathbf{w}\cdot\mathbf{x}+b=0
\]

This equation defines the boundary in terms of the model's weights and bias.

### 5.1 Linear Decision Boundaries

With two features, the decision boundary is a straight line:

\[
w_1x_1+w_2x_2+b=0
\]

Consider the following example:

\[
w_1=1,\quad w_2=1,\quad b=-3
\]

Substituting these values gives:

\[
x_1+x_2-3=0
\]

Therefore:

\[
x_1+x_2=3
\]

Points satisfying this equation lie on the decision boundary. Points on opposite sides receive different predicted classes, based on the sign of the linear expression.

With three features, the boundary is generally a plane in three-dimensional feature space. With more features, it becomes a higher-dimensional hyperplane.

## 6. Nonlinear Decision Boundaries

A linear decision boundary cannot adequately separate every dataset. Some classification problems contain circular, elliptical, or more complex patterns.

Logistic regression can model nonlinear boundaries by incorporating engineered features into its linear component.

For example, instead of using only \(x_1\) and \(x_2\), the model can use:

\[
x_1,\quad x_2,\quad x_1^2,\quad x_2^2
\]

The resulting model may be written as:

\[
z=w_1x_1+w_2x_2+w_3x_1^2+w_4x_2^2+b
\]

The decision boundary remains linear in the model's parameters, but it can be nonlinear in the original input features.

Additional terms, such as \(x_1x_2\), \(x_1^3\), or other polynomial combinations, can provide greater flexibility.

### Example: Circular Boundary

A model using the features \(x_1^2\) and \(x_2^2\) can learn a boundary of the form:

\[
x_1^2+x_2^2=r^2
\]

This equation describes a circle centered at the origin with radius \(r\).

By learning appropriate feature weights, logistic regression can approximate this type of classification pattern.

**Important:** More polynomial features increase model flexibility, but they can also increase the risk of overfitting. Feature selection and model evaluation are important when deciding which transformations to use.

## 7. Practical Implementation with Python

The following example demonstrates binary logistic regression using Scikit-learn.

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

# Input features and binary labels
X = np.array([
    [1.0, 2.0],
    [2.0, 1.0],
    [3.0, 3.0],
    [4.0, 4.0]
])

y = np.array([0, 0, 1, 1])

# Train the classifier
model = LogisticRegression()
model.fit(X, y)

# Predict probabilities for a new observation
new_sample = np.array([[3.5, 3.0]])

probabilities = model.predict_proba(new_sample)
prediction = model.predict(new_sample)

print("Class probabilities:", probabilities)
print("Predicted class:", prediction[0])
```

### Understanding the Code

- `LogisticRegression()` creates the classification model.
- `fit(X, y)` learns the weights and bias from the training data.
- `predict_proba()` returns the estimated probability for each class.
- `predict()` returns the predicted class label.

The example uses a small dataset to demonstrate the workflow. In real applications, model performance should be evaluated using suitable training and test datasets.

### Adding Polynomial Features

Scikit-learn can generate polynomial features before fitting logistic regression.

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import LogisticRegression

model = make_pipeline(
    PolynomialFeatures(degree=2, include_bias=False),
    StandardScaler(),
    LogisticRegression()
)

model.fit(X, y)
```

This pipeline creates polynomial features, scales them, and trains a logistic regression classifier.

For a production model, the polynomial degree and regularization settings should be selected using appropriate validation procedures. The pipeline also helps ensure that feature transformations are applied consistently during training and prediction.

## 8. Key Takeaways

- **Classification** predicts discrete categories, whereas regression predicts continuous numerical values.
- **Binary classification** predicts between two classes, commonly encoded as 0 and 1.
- Linear regression is not naturally suited to classification because its outputs are unbounded.
- **Logistic regression** combines a linear function with a sigmoid function to estimate class probabilities.
- The sigmoid function maps any real-valued input to a value between 0 and 1.
- A predicted probability is converted into a class label using a decision threshold.
- With a threshold of 0.5, the decision boundary satisfies \(\mathbf{w}\cdot\mathbf{x}+b=0\).
- Polynomial features allow logistic regression to represent nonlinear decision boundaries.
- Higher model flexibility can improve classification of complex patterns but may also lead to overfitting.
- Scikit-learn provides practical tools for training logistic regression models and building pipelines for feature engineering and scaling.
