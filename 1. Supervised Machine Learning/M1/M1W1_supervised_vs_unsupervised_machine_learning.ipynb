# Machine Learning — Supervised, Unsupervised Learning, and Jupyter Notebooks

## 1. Machine Learning

Machine learning (ML) is a field of computer science in which computers learn useful patterns or behaviors from experience and data rather than requiring every rule to be explicitly programmed.

A classical definition was given by **Arthur Samuel**, whose work on a checkers-playing program demonstrated that a computer could improve its performance through experience.

Instead of manually programming every strategy, the program played many games against itself and used that experience to improve its gameplay.

This illustrates one of the fundamental ideas of machine learning:

> A machine learning system can improve its behavior by learning from data or experience.

---

# 2. Major Machine Learning Approaches

Machine learning contains several learning paradigms. The two foundational approaches introduced here are:

* **Supervised learning**
* **Unsupervised learning**

Other important approaches include:

* Reinforcement learning
* Recommender systems

The key difference between supervised and unsupervised learning is whether the training data contains known output labels.

| Approach              | Input | Known Output | Main Objective            |
| --------------------- | ----- | ------------ | ------------------------- |
| Supervised learning   | X     | Y            | Learn to predict Y from X |
| Unsupervised learning | X     | None         | Discover structure in X   |

---

# 3. Supervised Learning

**Supervised learning** is a machine learning approach in which an algorithm learns a relationship between an input `X` and a known output `Y`.

The training dataset contains examples where the correct answer is already known.

Conceptually:

```text
Input X + Correct Output Y
            ↓
     Learning Algorithm
            ↓
       Learned Model
            ↓
New Input X → Predicted Y
```

The objective is to learn a function that can produce useful predictions when presented with new inputs for which the correct output is not yet known.

## 3.1 Labeled Data

The examples used for supervised learning are called **labeled data** because each input is associated with a known target or answer.

For example:

| Input X             | Output Y                    |
| ------------------- | --------------------------- |
| House size          | House price                 |
| Email content       | Spam / Not Spam             |
| Audio signal        | Text transcript             |
| Product image       | Defective / Not defective   |
| Vehicle sensor data | Position of another vehicle |

The algorithm uses these examples to learn the relationship between inputs and outputs.

---

# 4. Applications of Supervised Learning

Supervised learning can be applied to many prediction and decision problems.

### Spam Filtering

The input can be an email and the output can indicate whether the email is spam.

```text
Email → Spam / Not Spam
```

### Speech Recognition

The input is an audio signal and the target output is a corresponding text transcript.

```text
Audio → Text
```

### Machine Translation

The input is text in one language and the output is its translation into another language.

```text
Language A → Language B
```

### Online Advertising

A model can use information about a user, advertisement, and context to estimate whether the user will click the advertisement.

```text
User + Advertisement + Context → Click Prediction
```

### Automotive Applications

Supervised learning can be used with sensor data to estimate the position or behavior of surrounding vehicles.

```text
Sensor Data → Predicted Vehicle Position
```

### Manufacturing Inspection

Images of manufactured products can be used to train models that identify defective products.

```text
Product Image → Defective / Non-Defective
```

# 5. Two Main Types of Supervised Learning

Supervised learning problems can be broadly divided into:

1. **Regression**
2. **Classification**

The distinction is primarily based on the type of output that the model is expected to predict.

```text
Supervised Learning
       │
       ├── Regression
       │     └── Continuous numerical output
       │
       └── Classification
             └── Discrete categorical output
```

---

# 6. Regression

**Regression** is a supervised learning problem in which the target output is a **continuous numerical value**.

A classic example is house-price prediction.

The input could be:

```text
House size
```

and the target could be:

```text
House price
```

For example:

| House Size |    Price |
| ---------: | -------: |
|     100 m² | €250,000 |
|     120 m² | €290,000 |
|     150 m² | €350,000 |
|     200 m² | €460,000 |

The model learns from known examples and attempts to predict the price of a new house.

## 6.1 Learning a Prediction Function

A regression algorithm attempts to learn a function that maps the input to a numerical output:

```text
House Size → Learned Function → Predicted Price
```

Different mathematical functions can be used to represent the relationship between the input and output.

For example, the relationship could be approximated by a straight line or by a more complex curve.

The important characteristic is that the output can take values across a continuous numerical range.

---

# 7. Classification

**Classification** is a supervised learning problem in which the output belongs to a finite set of categories or classes.

For example, a medical model could classify a tumor as:

```text
0 → Benign
1 → Malignant
```

The model receives information about the tumor and predicts which category it belongs to.

Unlike regression, the output is not an arbitrary continuous numerical value.

## 7.1 Binary Classification

When there are two possible classes, the problem is called **binary classification**.

Example:

```text
Tumor → Benign / Malignant
```

Other examples include:

```text
Email → Spam / Not Spam
Product → Defective / Non-Defective
Transaction → Fraud / Not Fraud
```

---

# 8. Multi-Class Classification

Classification does not have to be limited to two categories.

A classification problem can contain multiple possible classes.

For example, instead of determining whether a tumor is simply benign or malignant, a system could classify a medical condition into several diagnostic categories.

Conceptually:

```text
Input
  ↓
Classification Model
  ↓
Class A
Class B
Class C
Class D
```

The output is still categorical even though more than two classes are possible.

---

# 9. Multiple Input Features

A supervised learning model does not have to rely on a single input.

Real-world problems commonly use **multiple features** to make a prediction.

For example, a tumor classification model might use:

* Tumor size
* Patient age
* Tumor shape
* Cell characteristics
* Other measured properties

Conceptually:

```text
Feature 1 ─┐
Feature 2 ─┤
Feature 3 ─┼──→ Classification Model → Predicted Class
Feature 4 ─┤
Feature 5 ─┘
```

Using multiple relevant features can provide the model with more information about the problem.

---

# 10. Decision Boundaries

Classification models can learn a **decision boundary** that separates different classes in the feature space.

For a simple two-feature problem, the model might receive:

```text
Feature 1 → Tumor Size
Feature 2 → Patient Age
```

The learned model can identify a boundary separating examples belonging to different classes.

Conceptually:

```text
             Feature 2
                 ↑
        Class A  |  Class A
                 |
-----------------+----------------→ Feature 1
                 |
        Class B  |  Class B
                 |
```

In real machine learning problems, the decision boundary can be much more complex and may involve many features.

---

# 11. Unsupervised Learning

**Unsupervised learning** is a machine learning approach where the algorithm receives input data but does not receive predefined output labels.

Conceptually:

```text
Input Data X
     ↓
Learning Algorithm
     ↓
Discover Patterns / Structure
```

The algorithm is not given the correct answer for each example.

Instead, it attempts to identify useful structure within the available data.

This makes unsupervised learning fundamentally different from supervised learning.

| Supervised                          | Unsupervised                             |
| ----------------------------------- | ---------------------------------------- |
| Uses labeled examples               | Uses unlabeled data                      |
| Has known target outputs            | No predefined target outputs             |
| Learns X → Y relationships          | Discovers structure in X                 |
| Prediction is usually the objective | Pattern discovery is often the objective |

---

# 12. Clustering

**Clustering** is one of the major forms of unsupervised learning.

A clustering algorithm attempts to group data points that are similar to one another.

Conceptually:

```text
Unlabeled Data
      ↓
  Clustering
      ↓
┌─────────┐
│ Cluster │
│    A    │
└─────────┘

┌─────────┐
│ Cluster │
│    B    │
└─────────┘

┌─────────┐
│ Cluster │
│    C    │
└─────────┘
```

The important characteristic is that the groups are not necessarily provided beforehand.

The algorithm attempts to discover them from the data.

---

# 13. Examples of Clustering

### News Articles

News articles can be grouped according to similarities in their content.

For example:

```text
News Articles
      ↓
   Clustering
      ↓
Technology
Sports
Politics
Finance
Science
```

The system does not necessarily need predefined labels for every article. It can discover groups based on similarities in the available data.

### Genetic Data

Clustering can also be applied to genetic measurements to identify groups of individuals with similar patterns of gene expression.

### Market Segmentation

Businesses can use clustering to group customers according to characteristics or behaviors.

For example:

```text
Customer Data
     ↓
 Clustering
     ↓
Group 1 → Similar customer behavior
Group 2 → Similar customer behavior
Group 3 → Similar customer behavior
```

This can help organizations understand different customer groups and tailor services to them.

---

# 14. Other Types of Unsupervised Learning

Clustering is only one form of unsupervised learning.

Other important categories include:

### Anomaly Detection

**Anomaly detection** attempts to identify observations that are unusual compared with the normal patterns in a dataset.

A major application is fraud detection.

For example:

```text
Normal Transactions
Normal Transactions
Normal Transactions
Unusual Transaction ← Potential Anomaly
Normal Transactions
```

The objective is to identify unusual behavior that may require further investigation.

### Dimensionality Reduction

**Dimensionality reduction** involves reducing the number of variables or dimensions used to represent data while attempting to preserve important information.

Large datasets can contain hundreds or thousands of features.

Dimensionality reduction can transform such data into a smaller representation:

```text
High-Dimensional Data
        ↓
Dimensionality Reduction
        ↓
Lower-Dimensional Representation
```

This can make complex datasets easier to analyze, visualize, or process.

# 15. Supervised vs. Unsupervised Learning

The difference can be illustrated through common applications.

| Problem                 | Learning Type | Reason                                                       |
| ----------------------- | ------------- | ------------------------------------------------------------ |
| Spam filtering          | Supervised    | Training examples have known spam/non-spam labels            |
| Diabetes diagnosis      | Supervised    | Training examples have known diagnostic outcomes             |
| House-price prediction  | Supervised    | Training examples contain known prices                       |
| News grouping           | Unsupervised  | Groups are discovered from article similarities              |
| Market segmentation     | Unsupervised  | Customer groups are discovered from data                     |
| Fraud anomaly detection | Unsupervised  | Unusual patterns can be identified without predefined labels |

The most important distinction is therefore not the application itself, but **how the learning problem is formulated and what information is available during training**.

# 16. Practical Machine Learning

Learning algorithms is only one part of machine learning.

A practical machine learning course also emphasizes how to **apply algorithms effectively to real-world problems**.

A successful machine learning system requires more than selecting an algorithm. Engineers must understand the problem, the available data, the expected output, and how the resulting system will be used.

The practical mindset can be summarized as:

```text
Machine Learning Algorithm
          +
Problem Understanding
          +
Data Understanding
          +
Practical Implementation
          ↓
Useful Machine Learning System
```

This is important because an algorithm that works well in an isolated experiment may not automatically produce a useful real-world system.

# 17. Jupyter Notebook

**Jupyter Notebook** is an interactive development environment used in the course to experiment with Python and machine learning.

It runs through a web-browser-based interface and allows code, explanations, results, and experiments to be kept together.

A notebook is divided into **cells**.

The two primary cell types introduced are:

### Markdown Cells

Markdown cells are used for:

* Explanations
* Notes
* Headings
* Documentation
* Mathematical descriptions

They allow the learner to combine technical explanations with executable code.

Code cells contain Python code that can be executed interactively.

The learner can:

1. Modify the code.
2. Run the cell.
3. Observe the result.
4. Modify the code again.
5. Run it again.

This makes Jupyter particularly useful for experimentation and learning.

````python
#1. Supervised Learning: Basic Regression
from sklearn.linear_model import LinearRegression

# Training data

X = [[100], [120], [150], [200]]
y = [250000, 290000, 350000, 460000]

# Create and train the model
model = LinearRegression()

model.fit(X, y)

# Predict and train the model

new_house = [[180]]
prediction = model.predict(new_house)

print(f"Predicted price: {prediction[0]:.2f}")
````

````python
# 2. Classification - Binary Classification
from sklearn.linear_model import LogisticRegression

# Tumor size 
X = [[2], [3], [4], [7], [8], [9]]

y = [0, 0, 0, 1, 1, 1]

model = LogisticRegression()
model.fit(X, y)

new_tumor = [[6]]
prediction = model.predict(new_tumor)

print(f"Malignant" if prediction[0] == 1 else "Bening")
````

````python
# 3. Classification - Multiple input features
from sklearn.linear_model import LogisticRegression

# Features:
# [tumor_size, patient_age]
X = [
    [2, 25],
    [3, 30],
    [4, 35],
    [7, 55],
    [8, 60],
    [9, 65]
]

# 0 = benign, 1 = malignant
y = [0, 0, 0, 1, 1, 1]

model = LogisticRegression()
model.fit(X, y)

# New patient
new_patient = [[6, 50]]

prediction = model.predict(new_patient)

print("Malignant" if prediction[0] == 1 else "Benign")
````

````python
# 4. Classification - Multi-class
from sklearn.linear_model import LogisticRegression

# Example features
X = [
    [1, 10],
    [2, 12],
    [3, 15],
    [7, 30],
    [8, 32],
    [9, 35],
    [15, 50],
    [16, 52],
    [17, 55]
]

# Three classes: 0, 1, 2
y = [0, 0, 0, 1, 1, 1, 2, 2, 2]

model = LogisticRegression()
model.fit(X, y)

new_sample = [[10, 40]]

prediction = model.predict(new_sample)

print(f"Predicted class: {prediction[0]}")
````

````python
# 5. Unsupervised Learning -  Cluestering
from sklearn.cluster import KMeans

# Unlabeled customer data
# [age, annual_spending]
X = [
    [20, 1000],
    [22, 1200],
    [25, 1500],
    [40, 5000],
    [42, 5200],
    [45, 5500]
]

# Create clustering model
model = KMeans(n_clusters=2, random_state=42, n_init=10)

# Learn the groups
model.fit(X)

# Cluster assigned to each customer
clusters = model.labels_

for customer, cluster in zip(X, clusters):
    print(f"Customer {customer} → Cluster {cluster}")
````

````python
# 6. Unsupervised Learning - Anomaly Detection

from sklearn.ensemble import IsolationForest

# Example transaction data
X = [
    [100],
    [120],
    [110],
    [130],
    [115],
    [125],
    [5000]  # Potential anomaly
]

model = IsolationForest(
    contamination=0.1,
    random_state=42
)

model.fit(X)

predictions = model.predict(X)

for transaction, result in zip(X, predictions):
    if result == -1:
        print(f"Anomaly detected: {transaction[0]}")
    else:
        print(f"Normal transaction: {transaction[0]}")
````

````python
# 7. Unsupervised Learning - Dimensionality Reduction
from sklearn.decomposition import PCA

# Dataset with 3 features
X = [
    [10, 20, 30],
    [11, 21, 31],
    [12, 22, 32],
    [50, 60, 70],
    [51, 61, 71]
]

# Reduce 3 dimensions to 2
pca = PCA(n_components=2)

X_reduced = pca.fit_transform(X)

print("Original shape:", len(X), "samples ×", len(X[0]), "features")
print("Reduced shape:", X_reduced.shape)

print("\nReduced data:")
print(X_reduced)
````
